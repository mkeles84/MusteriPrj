```
import os, shutil
from google.colab import drive, files

drive.mount('/content/drive')

KOK = '/content/drive/My Drive'

# Listele (klasörler sonda / ile gösterilir)
ogeler = sorted(os.listdir(KOK))
for i, ad in enumerate(ogeler):
    isaret = '/' if os.path.isdir(os.path.join(KOK, ad)) else ''
    print(f"{i:3d}  {ad}{isaret}")

# Seç
secim = int(input("\nİndirmek istediğin dosyanın numarası: "))
kaynak = os.path.join(KOK, ogeler[secim])

if os.path.isdir(kaynak):
    print("Bu bir klasör, içine gir ya da KOK değişkenini o klasöre ayarla.")
else:
    # 1) Colab sistemine kopyala
    hedef = os.path.join('/content', os.path.basename(kaynak))
    shutil.copy(kaynak, hedef)
    print("Colab'a kopyalandı:", hedef)

    # 2) PC'ne indir (istemiyorsan bu satırı sil)
    files.download(hedef)
```
